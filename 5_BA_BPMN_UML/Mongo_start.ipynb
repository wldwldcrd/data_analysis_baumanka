```python
import pymongo
import pandas as pd
```

```python
client = pymongo.MongoClient('mongodb://localhost:27017')
```

```python
collection = client.movie.movies
```

```python
mov = collection.find({"year":1994})
```

```python
df = pd.DataFrame({'Title':[],'Year':[],'Rating':[]})
j = 0
for i in mov:
   df.loc[j] = [i['title'],i['year'],i['rating']]
   j += 1
```
