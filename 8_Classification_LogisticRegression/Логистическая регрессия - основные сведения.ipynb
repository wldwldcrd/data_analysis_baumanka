```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

### Загружаем данные


```python
df = pd.read_csv('seattle-weather.csv')
```

### Основная информация о данных

```python
df.head()
```

```python
df['weather'].value_counts()
```

### Конвертация колонки с датой. Выделение признака

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df['month'] = df['date'].dt.month
```

```python
df['quarter'] = df['date'].dt.quarter
```

### Создание колонки с таргетом. Приведение ее к нужному типу

```python
df.loc[df['weather'] == 'fog', 'target'] = 1
```

```python
df.head()
```

```python
df['target'].value_counts()
```

```python
df['target'] = df['target'].astype('int16')
```

### Удаление ненужных колонок и разделение датасета

```python
df.head()
```

```python
df = df.drop(['weather','date'],axis=1)
```


### Логистическая регрессия : деление сета


```python
X = df.drop(['target'],axis=1)
Y = df['target']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

### Логистическая регрессия : Тренировка и прогноз

```python
X_train,X_test, Y_train,Y_test = train_test_split(X,Y, test_size=.15)
```

```python
model = LogisticRegression().fit(X_train, Y_train)
```

```python
pipe = Pipeline(
    [
        ('Scaler',StandardScaler()),
        ('LR',LogisticRegression())
    ]
).fit(X_train, Y_train)
```

### Оценка работы модели

```python
mean_absolute_error(Y_test, model.predict(X_test))
```

```python
r2_score(Y_test, model.predict(X_test))
```

```python
pipe.score(X_test, Y_test)
```
